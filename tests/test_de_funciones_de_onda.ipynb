# Test de Funciones de onda $\psi (x)$

Supondremos una función de onda típica senoidal y buscaremos normalizar, calcular la densidad de probabilidad, su transformada de Fourier, valores esperados y varianza, asi como graficar.

```python
import sympy as sp
import matplotlib.pyplot as plt

sp.init_printing()

x, p = sp.symbols("x p", real=True)
A, k, L, hbar= sp.symbols("A k L hbar", positive=True)
n = sp.symbols("n", integer=True)

psi = sp.sin(k*x)
display(sp.Eq(sp.Function("psi")(x), psi))
```

Vamos a normalizar y a calcular la densidad de probabilidad.

```python
integral = sp.integrate(psi**2, (x, -2, 2))
N = 1 / sp.sqrt(integral)
display(sp.Eq(sp.Symbol("N"), N.simplify()))
print("Función normalizada:")
display(sp.Eq(sp.Function("psi")(x), N.simplify()*psi))
```

```python
densidad = sp.conjugate(psi)*psi
display(sp.Eq(sp.Eq(sp.Abs(sp.Function("psi")(x))**2, sp.conjugate(sp.Function("psi")(x)) * sp.Function("psi")(x)), densidad, evaluate=False))
```

Ahora la Transformada de Fourier, del espacio de posiciones al de momentos. Veremos que nuestro primer ejemplo nos regresa, en general, una integral sin resolver, para observar que los cálculos funcionan usaremos otra función ejemplo típica, una función gaussiana.

```python
kernel = (1/sp.sqrt(2*sp.pi*hbar)) * sp.exp((-sp.I*p*x)/hbar)
phi = sp.integrate(psi*kernel, (x, -sp.oo, sp.oo)).simplify()
display(sp.Eq(sp.Function("phi")(p), sp.Integral(psi*kernel, (x, -sp.oo, sp.oo))))
display(sp.Eq(sp.Function("phi")(p), phi))
# Falta implementar como tratar estos casos de integrales no resueltas, deltas de Dirac, ...
```

```python
psi2 = sp.exp(-x**2)
phi2 = sp.integrate(psi2*kernel, (x, -sp.oo, sp.oo)).simplify()
display(sp.Eq(sp.Function("phi")(p), sp.Integral(psi2*kernel, (x, -sp.oo, sp.oo))))
display(sp.Eq(sp.Function("phi")(p), phi2))
```

Ahora calculamos los valores esperados de $x$ y $x^{2}$ y la varianza $\sigma^2 = <x^2> - <x>^2$

```python
valor_esperado_x = sp.integrate(sp.conjugate(psi) * x * psi, (x, -L, L)).simplify()
valor_esperado_x2 = sp.integrate(sp.conjugate(psi) * x**2 * psi, (x, -L, L)).simplify()
varianza = valor_esperado_x2 - valor_esperado_x**2

display(sp.Eq(sp.Eq(sp.Symbol("<x>"), sp.Integral(sp.conjugate(psi) * x * psi, (x, -L, L)), evaluate=False), valor_esperado_x, evaluate=False))
display(sp.Eq(sp.Eq(sp.Symbol("<x^{2}>"), sp.Integral(sp.conjugate(psi) * x**2 * psi, (x, -L, L)), evaluate=False), valor_esperado_x2, evaluate=False))
display(sp.Eq(sp.Symbol("sigma")**2, varianza.simplify(), evaluate=False))

varianza_especial = varianza.subs({L:(n*sp.pi)/(2*k)})
display(sp.Eq(sp.Symbol("sigma")**2, varianza_especial.simplify(), evaluate=False))
```
