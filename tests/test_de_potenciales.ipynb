Supondremos diversos potenciales y calcularemos la Fuerza asociada, puntos de equilibrio y estabilidad, límites en infinito, aproximación armónica y frecuencia natural de oscilación, puntos de retorno, entre otros.

## Potenciales continuos
Comencemos con un potenciales sencillos como uan cosntante, cuadrático, trigonomético y exponencial.

```python
import sympy as sp

sp.init_printing()

x = sp.symbols("x", real=True)
V0, k = sp.symbols("V_0 k", positive=True)
```

```python
V_constante = V0
V_cuadrático = sp.Rational(1,2)*k*x**2
V_trig = V0*sp.cos(k*x)
V_exp = V0*sp.exp(-x**2)

potenciales = [V_constante, V_cuadrático, V_trig, V_exp]

display(V_constante)
display(V_cuadrático)
display(V_trig)
display(V_exp)
```

Comencemos calculando la fuerza asociada, la cual es $F = -\frac{dV}{dx}$

```python
for potencial in potenciales:
    display(sp.Eq(sp.Derivative(potencial, x), potencial.diff(x)))
```

Ahora, vamos con los puntos de equilibrio y si estos son estables o inestables

```python
for potencial in potenciales:
    puntos_eq = sp.solveset(potencial.diff(x), x, domain=sp.S.Reals)
    estabilidad = potencial.diff(x, 2)
    if estabilidad.is_positive:
        estabilidad = "estables"
    else:
        estabilidad = "inestables"
    print("Los puntos de equilibrio: ")
    display(puntos_eq)
    print(f"son {estabilidad}\n")
```

Para el caso anterior, habrá que ajustar la estabilidad cuando tengamos un conjunto infinito de puntos de equilibrio. Por el momento pasemos a los limites en infinito.

```python
for potencial in potenciales:
    display(sp.Eq(sp.Limit(potencial, x, sp.oo), sp.limit(potencial, x, sp.oo)))
```
